# HANDLING MISSING DATA

```python
#HANDLING MISSING DATA  -> NaN (Missing data)
#creating data with missing values
import pandas as pd
import numpy as np
data = {
    "Customer":['A','B','C','D'],
    "Product": ["Milk","Bread","Butter","Milk"],
    "Price":[50,np.nan,40,50],
    "Quantity":[2,1,np.nan,1]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Detect missing values
#check entire DataFrame
print(df.isnull())
#count missing values
print(df.isnull().sum())
#Drop missing data
#df.dropna(inplace=True) #drop rows with missing data
#print(df)
df.dropna(axis=1,inplace=True) #drop columns with missing data
print(df)
```

```python
#Fill missing values
#df["Price"].fillna(df["Price"].mean(),inplace=True)
df["Price"]=df["Price"].fillna(df["Price"].mean())
print(df)
```

```python
#forward fill - takes previous value
df.fillna(method="ffill",inplace=True)
df
```

```python
#backward fill - takes next value
df.fillna(method="bfill",inplace=True)
df
```

```python
#check non - missing values
print(df.notnull())
print(df.notnull().sum())
df
```

```python
# Bivariate analysis -> Analysing relationships between TWO Variables
'''
Price     Quantity sold
10          100
20          80
30          60
40          40
Goal: Does price affect quantity sold?
'''
'''
# techniques to use
1) Scatter Plot  -> Visual Reationship
2) Correlation   -> Statistical Relationship
3) Covariance -> Direction of relationship
4) Regression   ->  Prediction
'''
'''
1) Scatter Plot-> Plot 2 variables
ex: x-axis -> Price
    y-axis -> Quantity

    If points go to Downward -> Negative relation
    If points go to Upward -> Positive relation
'''
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Price": [10,20,30,40],
    "Quantity":[100,80,60,40]
})
plt.scatter(df["Price"],df["Quantity"])
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.title("Scatter Plot")
plt.show()
#
```

```python
'''
2) Correlation
formula -> -1 <= r <= 1
correlation values:
 +1    Perfect positive correlation
 0     No correlation (no relation)
 -1    Perfect negative correlation
'''
print(df.corr())
```

```python
'''
3) Covariance
Positive -> Both Increase together
Negative -> One increase other decrease
'''
print(df.cov())
```

```python
'''
4) Regression(Prediction) -> Predict one variable using another
Predict quantity using price
Formula = y= mx+b
y = quantity
x = price
'''
```

```python
#Multivariate- Analysis relation between more than 2 variables
'''
Price   Discount   Advertisement   Sales
50        10%         1000          500
40        20%         2000          700
Goal: How price + Discount + Advertisement together affect sales
1) Multiple Regression
Equation: y = b0+ b1x1+ b2x2+b3x3
y = Sales
x1 =  price
x2 = Discount
x3 = Advertisement
formula -> -1 <= r <= 1
correlation values:
 +1    Perfect positive correlation
 0     No correlation (no relation)
 -1    Perfect negative correlation(perfect opposite movement)
'''
import pandas as pd

df = pd.DataFrame({
    "Price":[50,40,30],
    "Discount":[10,20,30],
    "Ads":[1000,2000,1500],
    "Sales":[500,700,650]})
print(df.corr())
'''
price incease -> sales decrease -> negative
discount increase -> sales increase -> positive
ads increase -> sales increase -> strong positive
price increase -> discount decrease -> perfect negative
'''
```
