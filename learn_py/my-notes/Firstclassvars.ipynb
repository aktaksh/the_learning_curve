Python function are same like first class fucntions





```python
def msg(name):
    return f"Hello {name}!"
f = msg  # assign func to var
print(f("Emma")) # call func as var
```

```python
#pass func as args
def msg(name):
    return f"Hello  dear  {name}"
def fnasarg(funasvar, name ):
    return funasvar(name)

print(fnasarg(msg, "Tiger"))
```
