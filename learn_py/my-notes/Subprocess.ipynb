Subprocess is module which is used to make systemcalls and store output for a specific proces
PIPE
call
run
stdout
capture_output


```python
import subprocess
#run command andget output
uptime = subprocess.run(["uptime"], capture_output=True , text=True).stdout.strip()
print(uptime)
#11:27  up 3 days,  1:04, 2 users, load averages: 2.97 2.52 2.73
#11:27  up 3 days,  1:04, 2 users, load averages: 3.30 2.63 2.76

#long output from long running process
proc = subprocess.Popen(["tail", "-f", "/var/log"], stdout=subprocess.PIPE, text=True)
print(proc)

#checkoutput -> get command output string
list_file = subprocess.check_output(["ls"], text=True)
print(list_file)
print("==================================")
#check_call()/ getstatusoutput -> check call output for non zero exit status for process
print(subprocess.check_call(["ls", "-l"]))

```
