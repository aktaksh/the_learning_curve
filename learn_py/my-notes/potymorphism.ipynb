```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meows")
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

```python
from abc import ABC, abstractmethod 
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal)

```

```python
#static method - method belongs to a class but does not need access to an object
class Calculator:

    @staticmethod
    def add(a,b):
        return a+b

    def multi(self,a,b):
        return a * b
        
print(Calculator.add(9,10))
m = Calculator()
print(m.multi(9,10))

```

```python
#Class method. - when object needs access to class itself receives cls automatically
class Server:
    platform = "Linux"

    @classmethod
    def show_platform(cls):
        return cls.platform

print(Server.show_platform())
```
