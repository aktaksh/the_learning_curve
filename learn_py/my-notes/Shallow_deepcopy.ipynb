diff in shallow and deep copy

```python
import copy
a  = [[1,2], [3,4]]
print(id([[1,2], [3,4]]))

print(id(a))
b = copy.copy(a)
print(id([[1,2], [3,4]]))

print(id(b))
c = copy.deepcopy(a)
print(id([[1,2], [3,4]]))
print(id(c))

a[0][0] = 99
print(a[0][0])
print(b[0][0])
print(c[0][0])
```
